```python
print("Dados únicos")
print(df.nunique())
print("-Dados nulos")
print(df.isnull().sum())
print("Porcentagem de dados nulos")
df.isnull().sum() / len(df) * 100
```

```python
#Histogramas de Velocidade do vento, direção do vento em graus, Temp(oC), Umid.Rel.(%), Pressão(mb) respectivamente
sns.histplot(data=df, x = 'Vel.VT(m/s)', color='red')
plt.show()

sns.histplot(data=df, x = 'Dir.VT(o)', color='blue')
plt.show()

sns.histplot(data=df, x = 'Temp(oC)', color='green')
plt.show()

sns.histplot(data=df, x = 'Umid.Rel.(%)', color='purple')
plt.show()

sns.histplot(data=df, x = 'Pressão(mb)', color='red')
plt.show()

# Histograma de PLUVIOSIDADE (mm)
bins_hist = np.arange(0, df['PLU(mm)'].max() + 0.3, 0.3)
sns.histplot(data=df, x='PLU(mm)', color='red', bins=bins_hist, binwidth=0.3)

#Plotando... 
plt.xlim(0, df['PLU(mm)'].max())
max_valor = df['PLU(mm)'].max()
plt.xticks(np.arange(0, max_valor + 0.3, 0.3))
plt.xticks(rotation=45, size=8)  
plt.tight_layout()
plt.show()
```

```python
#Boxplot, com LIM INFERIOR, Q1, Q2(MEDIANA), Q3, LIM SUPERIOR E OUTLIERS
#Pluviosidade, Vel Vento, Direção do vento, Temperatura, Umidade Relativa, Pressão
sns.boxplot(data=df, x = 'PLU(mm)')
plt.show()
sns.boxplot(data=df, x = 'Vel.VT(m/s)', color='red')
plt.show()
sns.boxplot(data=df, x = 'Dir.VT(o)', color='blue')
plt.show()
sns.boxplot(data=df, x = 'Temp(oC)', color='green')
plt.show()
sns.boxplot(data=df, x = 'Umid.Rel.(%)', color='purple')
plt.show()
sns.boxplot(data=df, x='Pressão(mb)', color='red')
```

```python
#Heatmap 
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', center=0)
```

```python
#Pairplot
sns.pairplot(df)
```
